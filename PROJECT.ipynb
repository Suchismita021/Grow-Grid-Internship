The main objective of this project is to understand the prescription patterns through Exploratory Data Analysis (EDA) and build a machine learning classification model that can predict the appropriate drug category for a patient.

```python
#44 Are there any unusually high or low Na_to_K values?
# Identify them using the IQR method.
Q1 = train['Na_to_K'].quantile(0.25)

Q3 = train['Na_to_K'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR

upper_limit = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
```
