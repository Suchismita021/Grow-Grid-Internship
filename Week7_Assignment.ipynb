```python
#1. How does average monthly demand vary across different product categories?
df.groupby('category')['avg_monthly_demand'].mean().sort_values(ascending=False)
```

```python
#2. Which product categories show the highest demand variability?
df.groupby('category')['demand_variability'].mean().sort_values(ascending=False)
```

```python
#3. Which warehouse zones handle products with the highest average demand?
df.groupby('warehouse_zone')['avg_monthly_demand'].mean().sort_values(ascending=False)
```

```python
#4. Which products have the longest average supplier lead times?
df.groupby('item_id')['avg_lead_time_days'].mean().sort_values(ascending=False).head(10)
```

```python
#5. Is high lead-time variability associated with more supplier delay incidents?
df[['lead_time_variability', 'supplier_delay_incidents']].corr()
```

```python
#6. Which products have the highest stock turnover ratios?
df.groupby('item_id')['stock_turnover_ratio'].mean().sort_values(ascending=False).head(10)
```

```python
#7. Which products remain in inventory for the longest number of days?
df.groupby('item_id')['days_in_inventory'].mean().sort_values(ascending=False).head(10)
```

```python
#8. Is there a relationship between stock turnover ratio and days in inventory?
df[['stock_turnover_ratio', 'days_in_inventory']].corr()
```

```python
#9. Which products experience the highest stockout frequency?
df.groupby('item_id')['stockout_frequency'].sum().sort_values(ascending=False).head(10)
```

```python
#10. Are products with high demand variability more likely to experience stockouts?
df[['demand_variability', 'stockout_frequency']].corr()
```

```python
#11. Which products experience frequent backorders?
df.groupby('item_id')['backorder_frequency'].sum().sort_values(ascending=False).head(10)
```

```python
#12. Is backorder frequency associated with stockout frequency?
df[['backorder_frequency', 'stockout_frequency']].corr()
```

```python
#13. Which products have the highest reorder-point breaches?
df.groupby('item_id')['reorder_point_breaches'].sum().sort_values(ascending=False).head(10)
```

```python
#14. Do products with frequent reorder-point breaches also have high monthly reorder counts?
df[['reorder_point_breaches', 'monthly_reorder_count']].corr()
```

```python
#15. Which products have the highest supplier delay incidents?
df.groupby('item_id')['supplier_delay_incidents'].sum().sort_values(ascending=False).head(10)
```

```python
#16. Is supplier delay associated with increased stockouts or backorders?
df[['supplier_delay_incidents', 'stockout_frequency', 'backorder_frequency']].corr()
```

```python
#17. Which products maintain significantly higher safety stock relative to their current stock?
df['safety_stock_ratio'] = df['safety_stock'] / df['current_stock']
```

```python
df.groupby('item_id')['safety_stock_ratio'].mean().sort_values(ascending=False).head(10)
```

```python
#18. Which products have high excess-stock levels?
df.groupby('item_id')['excess_stock_level'].mean().sort_values(ascending=False).head(10)
```

```python
#19. Is excess stock associated with high days in inventory?
df[['excess_stock_level', 'days_in_inventory']].corr()
```

```python
#20. Which products have the highest inventory value?
df.groupby('item_id')['inventory_value'].sum().sort_values(ascending=False).head(10)
```

```python
#21. Which products have the highest holding cost per unit?
df.groupby('item_id')['holding_cost_per_unit'].mean().sort_values(ascending=False).head(10)
```

```python
#22. Do expensive products tend to have lower or higher stock turnover?
df[['unit_cost', 'stock_turnover_ratio']].corr()
```

```python
#23. Which products have the highest return rates?
df.groupby('item_id')['return_rate'].mean().sort_values(ascending=False).head(10)
```

```python
#24. Which products have the highest damage rates?
df.groupby('item_id')['damage_rate'].mean().sort_values(ascending=False).head(10)
```

```python
#25. Are high return and damage rates concentrated in particular categories or warehouse zones?
df.groupby('category')[['return_rate', 'damage_rate']].mean().sort_values('return_rate', ascending=False)
```

```python
df.groupby('warehouse_zone')[['return_rate', 'damage_rate']].mean().sort_values('return_rate', ascending=False)
```

```python
#26. Which products have the highest warehouse demand pressure index?
df.groupby('item_id')['warehouse_demand_pressure_index'].mean().sort_values(ascending=False).head(10)
```
